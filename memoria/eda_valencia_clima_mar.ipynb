IDEA

Al principio no tenía claro sobre qué hacer el EDA, mi primera idea fue encontrar un DataBase de los jugadores de futbol de la liga española y agruparlos por características físicas y ver si la relación entre altura y peso guardaba relación con el tipo de lesiones que podían padecer, pero los datos que vi por internet no me convencieron y decidí tirar por los datos meteorológicos porque sabía había múltiples formas de obtener este tipo de datos de forma libre y gratuita bajo la premisa de discernir si la temperatura del Mar Mediterráneo frente a la costa de Valencia tenía efectos en la creación de las DANAS y si  estos fenómenos eran cada vez más frecuentes, tomando datos de los últimos 5 años.

API AEMET

No me veía con capacidad para intentar web scraping con los plazos que manejábamos y descargarme un DataSet de kaggle me parecía demasiado sencillo, así que tiré por el camino de enmedio y intenté obtener los datos haciendo peticiones a la api de Aemet (Open Meteo) a través de python. Esto me ayudó a entender mejor el temario de Apis que hasta ese momento no había conseguido interiorizar.

COPERNICUS

Conseguir los datos sobre la temperatura del mar mediterraneo no fue tan sencillo, después de buscar en varios portales, me encontré con El Servicio de Vigilancia Medioambiental Marina de Copernicus (CMEMS)junto con unos tutoriales sobre como descargar información de su web, y es que este servicio de observción de la Tierra de la Unión Europea cuenta con decenas de mapas interactivos plagados de información sobre el mar, pero la estraacción de estos datos no es trivial así que después de conseguir descargar un archivo con la información que necesitaba, la temperatura de la superficie del mediterráneo, se descargó un archivo en un formato raro (.nc) y tuve que tirar mano de la IA para transformarlo en algo que mi ordenador puediera leer.

PRIMEROS PASOS

Una vez descargados ambos archivos me dediqué a convertirlos en DataFrames para poder mostrarlos por pantalla, ver si tenían nulos y, comprender el tipo de datos con los que contaba. Cambié los formatos de las fechas de strings a Datetime y limpié el resto de datos para que fueran números. Había 5 Nan en columnas referentes a los datos de precipitaciones así que decidí rellenarlos con la media del resto de elementos de la columna que tenían valores similares en las columnas de precipitación que sí estaban. Tras este proceso, junté los dos Dataframes en uno solo para poder trabajar más cómodamente y lo guardé como .csv en la carpeta data>processed para poder trabajar posteriormente sin tener que repetir el proceso.

TRATAMIENTO DE LOS DATOS

Tras un primer proceso de tratamiento en el que conseguí visualizar las columnas en múltiples tablas para poder apreciar la forma de los datos, tome la decisión de aumentar el rango de los mismos. Lo que en un primer momento iban a ser 5 años, terminron siendo 35 y con estos nuevos datos vi la posibilidad de añadir nuevas columnas al Dataframe que ya teníamos así que de nuevo volví al proceso de limpieza.

LIMPIEZA DE DATOS

Estas nuevas columnas añadidas (presión atmosférica, humedad relativa, velocidad del viento...) presentaban una nueva oportunidad de encontrar más patrones y correlaciones; sin embargo también suponían un problema puesto que aproximadamente la mitad de estos valores eran nulos (NaN). Así que tuve que tomar la decisión de prescindir de ellas, no sin pena. Sin embargo, ya me había hecho a la idea de contar con esas columnas por lo que decidí buscar en internet dónde encontrar esos datos que me faltaban.
