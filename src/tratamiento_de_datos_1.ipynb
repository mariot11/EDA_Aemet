```python
df_meteo['año'] = df_meteo.index.year

# 2. Hacer todo el cálculo en una sola línea de código con .agg()
resumen_anual = df_meteo.groupby('año').agg({
    'temp_med_mar': ['max', 'mean'],
    'temp_med_mes': ['max', 'mean'],
    'temp_abs_max': ['max', 'mean'],
    'temp_abs_min': ['min', 'mean'],
    'dias_temp>30': ['sum', 'mean'],
    'total_precip_mes': ['sum', 'mean'],
    'precip_max': ['max', 'mean'],
    'dias_precip>10mm': ['sum', 'mean'],
    'dias_precip>30mm': ['sum', 'mean'],
    'dias_lluvia': ['sum', 'mean'],
    'dias_granizo': ['sum', 'mean'],
    'dias_tormenta': ['sum', 'mean'],
    'total_precip_mes': ['sum', 'mean']
})

# Opcional (muy recomendado): Aplanar el índice multinivel de las columnas resultante
resumen_anual.columns = ['_'.join(col).strip() for col in resumen_anual.columns.values]

# Aplicar el renombramiento limpio de todas las columnas
resumen_anual = resumen_anual.rename(columns={
    # Temperatura del Mar y Mes
    'temp_med_mar_max': 'temp_med_max_mar',
    'temp_med_mar_mean': 'temp_med_mar',
    'temp_med_mes_mean': 'temp_med',
    
    # Temperaturas Extremas (Absolutas)
    'temp_abs_max_max': 'temp_abs_max',
    'temp_abs_max_mean': 'temp_max_mes_med',
    'temp_abs_min_min': 'temp_abs_min',
    'temp_abs_min_mean': 'temp_min_med',
    
    # Días cálidos
    'dias_temp>30_sum': 'dias_temp>30',
    'dias_temp>30_mean': 'dias_temp>30_med',
    
    # Precipitaciones (siguiendo tu ejemplo)
    'total_precip_mes_sum': 'total_precip_anio',
    'total_precip_mes_mean': 'precip_med_anio',
    'precip_max_max': 'precip_max_abs',
    'precip_max_mean': 'precip_max_mes_med',
    
    # Días de lluvia intensa
    'dias_precip>10mm_sum': 'dias_precip_10mm',
    'dias_precip>10mm_mean': 'dias_precip_10mm_med',
    'dias_precip>30mm_sum': 'dias_precip_30mm',
    'dias_precip>30mm_mean': 'dias_precip_30mm_med',
    
    # Fenómenos (Lluvia, Granizo, Tormenta)
    'dias_lluvia_sum': 'dias_lluvia',
    'dias_lluvia_mean': 'dias_lluvia_med',
    'dias_granizo_sum': 'dias_granizo',
    'dias_granizo_mean': 'dias_granizo_med',
    'dias_tormenta_sum': 'dias_tormenta',
    'dias_tormenta_mean': 'dias_tormenta_med'
})

resumen_anual.plot(
    kind="line", 
    subplots=True, 
    layout=(-1, 2), 
    figsize=(14, 18), 
    marker="o",      # Pone un puntito en cada año para que se vea claro
    grid=True,       # Añade cuadrícula para facilitar la lectura
    sharex=True      # Comparte el eje X (los años) para no repetir números
)

plt.suptitle("Evolución histórica de las variables entre 2021-2025 (Valencia)", fontsize=16, y=1)
plt.tight_layout()
plt.show()
```
