```python
df_meteo['año'] = df_meteo.index.year

# 2. Hacer todo el cálculo en una sola línea de código con .agg()
resumen_anual = df_meteo.groupby('año').agg({
    # Temperatura: Nos quedamos con la media global y los extremos absolutos del año
    'temp_med_mar (Cº)': ['mean'],          # Temperatura media del mar
    'temp_med_mes (Cº)': ['mean'],          # Temperatura media anual
    'temp_abs_max (Cº)': ['max'],           # Récord de calor del año (pico más alto)
    'temp_abs_min (Cº)': ['min'],           # Récord de frío del año (pico más bajo)
    
    # Días cálidos y Precipitaciones (Acumulados anuales)
    'dias_temp>30': ['sum'],                # Total de días calurosos al año
    'total_precip_mes (mm)': ['sum'],       # Precipitación total acumulada al año
    'precip_max (mm)': ['max'],             # Mayor chaparrón en un solo día del año
    
    # Conteo de días de lluvia y fenómenos (Totales anuales)
    'dias_precip>10mm': ['sum'],
    'dias_precip>30mm': ['sum'],
    'dias_lluvia': ['sum'],
    'dias_granizo': ['sum'],
    'dias_tormenta': ['sum'],
    
    # Humedad, Presión y Viento (Métricas clave simplificadas)
    'humedad_med (%)': ['mean'],            # Humedad media anual
    'presion_med (hPa)': ['mean'],          # Presión atmosférica media anual
    'presion_min (hPa)': ['min'],           # Presión mínima (detección de borrascas profundas)
    'viento_vel_max (km/h)': ['max'],       # Racha máxima de viento del año
    'viento_vel_med (km/h)': ['mean']
})


# Aplanar el índice multinivel de las columnas resultante
resumen_anual.columns = ['_'.join(col).strip() for col in resumen_anual.columns.values]

columnas_decimales = [
    'temp_med_mar (Cº)_mean',
    'temp_med_mes (Cº)_mean',
    'humedad_med (%)_mean',
    'presion_med (hPa)_mean',
    'viento_vel_med (km/h)_mean'
]

for col in columnas_decimales:
    if col in resumen_anual.columns:
        # print(col)
        resumen_anual[col] = np.round(resumen_anual[col], decimals=1)

# Aplicar el renombramiento limpio de todas las columnas
resumen_anual = resumen_anual.rename(columns={
    'temp_med_mar (Cº)_mean': 'temp_mar_media (Cº)',
    'temp_med_mes (Cº)_mean': 'temp_media_anual (Cº)',
    'temp_abs_max (Cº)_max': 'temp_max_absoluta (Cº)',
    'temp_abs_min (Cº)_min': 'temp_min_absoluta (Cº)',
    
    'dias_temp>30_sum': 'dias_temp_>30',
    'total_precip_mes (mm)_sum': 'precipitacion_total (mm)',
    'precip_max (mm)_max': 'precip_max_diaria (mm)',
    
    'dias_precip>10mm_sum': 'dias_lluvia_>10mm',
    'dias_precip>30mm_sum': 'dias_lluvia_>30mm',
    'dias_lluvia_sum': 'dias_lluvia_total',
    'dias_granizo_sum': 'dias_granizo_total',
    'dias_tormenta_sum': 'dias_tormenta_total',
    
    'humedad_med (%)_mean': 'humedad_media (%)',
    'presion_med (hPa)_mean': 'presion_media (hPa)',
    'presion_min (hPa)_min': 'presion_minima_anual (hPa)',
    'viento_vel_max (km/h)_max': 'viento_racha_maxima (km/h)',
    'viento_vel_med (km/h)_mean': 'viento_velocidad_media (km/h)'
})

resumen_anual.to_parquet("../data/processed/datos_meteo_Valencia_91-25(year).parquet")

resumen_anual.plot(
    kind="line", 
    subplots=True, 
    layout=(-1, 2), 
    figsize=(14, 24), 
    marker="o",      # Pone un puntito en cada año para que se vea claro
    grid=True,       # Añade cuadrícula para facilitar la lectura
    sharex=False      # Comparte el eje X (los años) para no repetir números
)

plt.suptitle("Evolución histórica de las variables entre 1991-2025 (Valencia)", fontsize=16, y=1)
plt.tight_layout()
plt.show()
```
